## Tâche 10 — Comparer

Une solution est une fonction `texte → liste de (debut, fin, categorie)`. Chaque outil a ses propres noms de catégories : il faut les faire correspondre à celles de `verite`.

```python
def point_de_comparaison(texte):
    return [(m.start(), m.end(), "EMAIL") for m in re.finditer(r"[\w.+-]+@[\w-]+\.\w+", texte)]


evaluer(point_de_comparaison, avis, verite)
```
